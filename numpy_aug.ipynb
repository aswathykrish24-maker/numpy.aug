```python
import numpy as np

#create a 4x4 matrix
z=np.array([[1,2,3,4],[3,2,1,5],[9,8,7,6],[4,5,3,2]])
print(z)
print(z.ndim)
print(z.size)
print(z.shape)
print(z[0])
print(z[2])
print(z[1,1])
print(z[1][1])
print(z[0:,1]) # to print 2nd indexing
print(z[0:,1:3]) # to print first and 2nd column
print(z[2,1:3]) # coma ku munne ullathu ethu row after column column
print(z[1:3,1:3])
print(z[0:,0:])# to print the whole matrix
```

```python
#create a 4x3 matrix with float values
z=np.array([[3.1,4.5,7.7,3.5],[2.9,6.3,5.5,8.5],[1.6,2.7,3.7,8.2]])
#floor
print(np.floor(z))
print(np.ceil(z))
print(np.round(z)) #nearest even
```

```python
z=np.array([[3.1,4.5,7.7,3.5],[2.9,6.3,5.5,8.5],[1.6,2.7,3.7,8.2]])
print(z.max())
print(z.min())
print(z.sum())
```

```python
z=np.array([[3.1,4.5,7.7,3.5],[2.9,6.3,5.5,8.5],[1.6,2.7,3.7,8.2]])
print(z.max(axis=1)) # represent rows in numpy but in pandas its column
print(z.max(axis=0)) # represent column in numpy
print(np.sqrt(z))
print(np.sin(z))
print(np.cos(z))
print(np.tan(z))



```

```python
z=np.array([[3.1,4.5,7.7,3.5],[2.9,6.3,5.5,8.5],[1.6,2.7,3.7,8.2]])
print(z)

print(np.sort(z)) # sort in row wise and asc order
print(np.sort(z,axis=0)) # sort in column wise and desc order
print(np.argsort(z)) # index based sort
```
